```python
import pandas as pd
import geopandas as gpd
from trailcalc import map # takes bbox and feature layers payload

#https://observablehq.com/@rdmurphy/u-s-state-bounding-boxes
CO_BBOX = {'WEST':-109.05919619986199,
           'SOUTH': 36.99275055519555,
           'EAST':-102.04212644366443,
           'NORTH': 41.00198213121131
           }

EAST_STEP = (CO_BBOX['EAST'] - CO_BBOX['WEST'])/16
NORTH_STEP = (CO_BBOX['NORTH'] - CO_BBOX['SOUTH'])/16


CELL_SOUTH_BOUND = CO_BBOX['SOUTH']
CELL_NORTH_BOUND = CO_BBOX['SOUTH'] + NORTH_STEP
CELL_WEST_BOUND = CO_BBOX['WEST']
CELL_EAST_BOUND = CO_BBOX['WEST'] + EAST_STEP


bbox_batches = {}
for i in range(16):
    bbox_batches[i] = []
    CELL_WEST_BOUND = CO_BBOX['WEST']
    CELL_EAST_BOUND = CO_BBOX['WEST'] + EAST_STEP
    for j in range(16):
        bbox_batches[i].append([CELL_NORTH_BOUND, CELL_SOUTH_BOUND, CELL_EAST_BOUND, CELL_WEST_BOUND])
        CELL_EAST_BOUND += EAST_STEP
        CELL_WEST_BOUND += EAST_STEP

    CELL_NORTH_BOUND += NORTH_STEP
    CELL_SOUTH_BOUND += NORTH_STEP

trail_data = gpd.GeoDataFrame({'geometry' : [], 'total_trail_mileage' : [], 'trail_density' : []})
```

```python


feature_layers_payload = {
    'highways': {
        'highway': ['motorway', 'trunk']
    },
    'roads': {
        'highway': ['primary', 'secondary', 'tertiary']
    },
    'streets': {
        'highway': ['residential', 'unclassified']
    },
    'trails': {
        'highway': ['path', 'footway']
    },
    'parks': {
        'leisure': ['park', 'nature_reserve'],
        'boundary': ['protected_area'],
        'landuse': ['grass'],
        'natural': ['wood']
    },
    'water': {
        'water': ['river', 'pond', 'lake', 'reservoir'],
        'waterway': ['river', 'canal'],
        'natural': ['water']
    },
    'buildings': {
        'building': True
    }
}

for bbox in bbox_batches[15]:
    print(bbox)
    projected_data = map.create_trail_mileage_map(bbox, feature_layers_payload)
    print(projected_data)
    try:
        projected_data_gpd = gpd.GeoDataFrame({'geometry' : [bbox], 'total_trail_mileage' : [projected_data['trail_miles']], 'trail_density' : [projected_data['trail_density_per_mile']]})
        trail_data = pd.concat([trail_data, projected_data_gpd], ignore_index=True)
    except:
        continue
    
print("Done!")
```

```python
from shapely.geometry import Polygon


trail_data['geometry'] = trail_data['geometry'].apply(lambda coords: Polygon([
    (coords[2], coords[0]),  # (west_lon, north_lat)
    (coords[3], coords[0]),  # (east_lon, north_lat)
    (coords[3], coords[1]),  # (east_lon, south_lat)
    (coords[2], coords[1]),  # (west_lon, south_lat)
]))

# Now create a GeoDataFrame
trail_data_gdf = gpd.GeoDataFrame(trail_data, geometry='geometry')
```

```python
import matplotlib.pyplot as plt

# Plotting the GeoDataFrame with a title
ax = trail_data_gdf.plot(
    column="total_trail_mileage", 
    scheme="Natural_Breaks", 
    k=9, 
    cmap='Greens', 
    linewidth=0, 
    legend=True, 
    legend_kwds={'bbox_to_anchor': (1.4, 1.025)}
)

# Add a title to the plot
ax.set_title('Heatmap of Colorado Trails', fontsize=16)

# Show the plot
plt.show()
```
